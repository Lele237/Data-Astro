```python
#dicionário de planetas

planetas = {
    "Terra" : {
        "massa" : 1.0,
        "raio" : 6351,
        "distancia_sol" : 149.6
    },

    "Marte" : {
        "massa" : 0.107,
        "raio" : 3389.5,
        "distancia_sol" : 227.9
    },
    "Vênus" : {
        "massa": 0.815,
        "raio" : 6051.8,
        "distancia_sol": 108.2
    },
    "Mercúrio" : {
        "massa": 0.055,
        "raio" : 2439.7,
        "distancia_sol": 57.9
    },

      "Júpiter": {
        "massa": 317.8,
        "raio": 69911,
        "distancia_sol": 778.5
    },

    "Saturno": {
        "massa": 95.2,
        "raio": 58232,
        "distancia_sol": 1432
    },

    "Urano": {
        "massa": 14.5,
        "raio": 25362,
        "distancia_sol": 2867
    },

    "Netuno": {
        "massa": 17.1,
        "raio": 24622,
        "distancia_sol": 4515
    }
}
```

```python
while True:
  print("===========SISTEMA SOLAR===========")
  print("1-Consultar Planeta")
  print("2- Comparar dois planetas")
  print("3- Mostrar Planeta mais massivo")
  print("4- Mostrar Planeta mais distante do Sol")
  print("5- Mostrar todos os planetas")
  print("0-sair")

  opcao = input("Escolha uma opção:")

  if opcao == "1":
    consultar_planeta(planetas)
    break
  elif opcao == "2":
    comparar_planetas(planetas)
    break
  elif opcao == "3":
    planeta_mais_massivo(planetas)
    break
  elif opcao == "4":
    planeta_mais_distante_sol(planetas)
    break
  elif opcao == "5":
    mostrar_todos_os_planetas(planetas)
    break
  elif opcao == "0":
    print("sair")
    break
  else:
    print("opção inválida")
```

```python
#Consultar planeta

def consultar_planeta(planetas):
  nome= input("Digite o nome de um planeta: ")

  if nome in planetas:
    planeta = planetas[nome]

    print(f"\nPlaneta: {nome}")

    print(f"Massa: {planeta['massa']} massas terrestres")

    print(f"Raio: {planeta['raio']} km ")

    print(f"Distancia do Sol: {planeta['distancia_sol']}")

  else:
    print("planeta não encontrado")


```

```python
#Comparar Planetas

def comparar_planetas(planetas):
  nome1 = input("Digite o primeiro planeta: ")
  nome2 = input("Digite o segundo planeta: ")


  if nome1 not in planetas or nome2 not in planetas or nome2 not in planetas:
    print("Um dos Planetas não foi encontrado encontrado")
    return

  planeta1 = planetas[nome1]
  planeta2 = planetas[nome2]

  print("========COMPARAÇÂO========")

  if planeta1 ["massa"] > planeta2 ["massa"]:
    print(f"{nome1} possui maior massa")
  else:
    print(f"{nome2} possui maior massa")

  if planeta1["raio"] > planeta2 ["raio"]:
    print(f"{nome1} possui maior raio")
  else:
    print(f"{nome2} possui maior raio")

  if planeta1 ["distancia_sol"] > planeta2 ["distancia_sol"]:
    print(f"{nome1} está mais distante do sol")
  else:
    print(f"{nome2} está mais distante do sol")
```

```python
#Planeta mais massivo
def planeta_mais_massivo(planetas):
  maior = None

  for nome in planetas:
    if maior is None:
      maior = nome

    elif planetas [nome] ["massa"] > planetas [maior] ["massa"]:
      maior = nome
  print(f"O planeta mais massivo é {maior}")
```

```python
#Planeta mais distante do sol
def planeta_mais_distante_sol(planetas):
  maior = None

  for nome in planetas:
    if maior is None:
      maior = nome

    elif planetas [nome] ["distancia_sol"] > planetas [maior] ["distancia_sol"]:
      maior = nome
  print(f"O planeta mais distante do sol é: {maior}")
```

```python
#Mostrar todos os planetas
def mostrar_todos_os_planetas(planetas):
  print("==========PLANETAS==========")
  for nome in planetas:
    print(nome)
```

```python
import pandas as pd

df = pd.DataFrame(planetas).T
df.index.name="planetas"
df = df.reset_index()
```
